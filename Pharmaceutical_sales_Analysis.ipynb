```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
Data = {"Drug": ["Paracetamol", "Amoxicillin", "Ibuprofen"],
"Price": [5,12,7],
"quantity_sold": [120, 80, 90]}
df = pd.DataFrame(Data)
df
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.shape)
```

```python
df["Revenue"] = df["Price"] * df["quantity_sold"]
```

```python
total_revenue = np.sum(df["Revenue"])
print("Total Revenue:", total_revenue)
```

```python
best_selling_drug = df.loc[df["quantity_sold"].idxmax()]
print(best_selling_drug)
```

```python
import matplotlib.pyplot as plt
plt.bar(df['Drug'], df['quantity_sold'], color= ['#4CAF50', '#008CBA', '#FF9800'])
plt.title('Pharmaceutical Sales Analysis - quantity Sold')
plt.xlabel('Drug')
plt.ylabel('quantity Sold')
plt.show()
```
